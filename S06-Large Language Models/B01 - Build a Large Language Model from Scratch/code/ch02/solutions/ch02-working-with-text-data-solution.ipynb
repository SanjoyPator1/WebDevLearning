- Putting it now all together into a tokenizer class

````python
class SimpleTokenizerV1:
    """A naive whitespace-and-punctuation tokenizer.

    Three methods to implement:

    __init__(self, vocab):
        Take a vocabulary dictionary that maps each token string to an
        integer id. Build a forward mapping (string -> id) and an inverse
        mapping (id -> string) as attributes on self so encode and decode
        can look up either direction.

    encode(self, text):
        Split the input text on whitespace and the common punctuation
        characters (comma, period, colon, semicolon, question mark,
        underscore, exclamation mark, quote, parenthesis, apostrophe,
        and the two-character double-dash). Strip whitespace off each
        resulting token and drop the empty strings. Look up every
        remaining token in the forward mapping and return the list of
        integer ids.

    decode(self, ids):
        Look up every id in the inverse mapping and join the resulting
        strings with a single space. Then clean up the spacing around
        punctuation -- any sequence of whitespace immediately before one
        of the punctuation marks should be removed so the text reads
        naturally. Return the cleaned text.
    """
    # YOUR CODE HERE
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.?!"()\'\\])', r'\1', text)
        return text
````

- The `encode` function turns text into token IDs
- The `decode` function turns token IDs back into text

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/08.webp?123" style="max-width: 100%; height: auto;">

- We can use the tokenizer to encode (that is, tokenize) texts into integers
- These integers can then be embedded (later) as input of/for the LLM

````python
tokenizer = SimpleTokenizerV1(vocab)

text = """"It's the last he painted, you know," 
           Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
````

- We can decode the integers back into text

````python
tokenizer.decode(ids)
````

````python
tokenizer.decode(tokenizer.encode(text))
````

## 2.4 Adding special context tokens

- It's useful to add some "special" tokens for unknown words and to denote the end of a text

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/09.webp?123" style="max-width: 100%; height: auto;">

- Some tokenizers use special tokens to help the LLM with additional context
- Some of these special tokens are
  - `[BOS]` (beginning of sequence) marks the beginning of text
  - `[EOS]` (end of sequence) marks where the text ends (this is usually used to concatenate multiple unrelated texts, e.g., two different Wikipedia articles or two different books, and so on)
  - `[PAD]` (padding) if we train LLMs with a batch size greater than 1 (we may include multiple texts with different lengths; with the padding token we pad the shorter texts to the longest length so that all texts have an equal length)
- `[UNK]` to represent words that are not included in the vocabulary

- Note that GPT-2 does not need any of these tokens mentioned above but only uses an `<|endoftext|>` token to reduce complexity
- The `<|endoftext|>` is analogous to the `[EOS]` token mentioned above
- GPT also uses the `<|endoftext|>` for padding (since we typically use a mask when training on batched inputs, we would not attend padded tokens anyways, so it does not matter what these tokens are)
- GPT-2 does not use an `<UNK>` token for out-of-vocabulary words; instead, GPT-2 uses a byte-pair encoding (BPE) tokenizer, which breaks down words into subword units which we will discuss in a later section



- We use the `<|endoftext|>` tokens between two independent sources of text:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/10.webp" style="max-width: 100%; height: auto;">

- Let's see what happens if we tokenize the following text:

- To deal with such cases, we can add special tokens like `"<|unk|>"` to the vocabulary to represent unknown words
- Since we are already extending the vocabulary, let's add another token called `"<|endoftext|>"` which is used in GPT-2 training to denote the end of a text (and it's also used between concatenated text, like if our training datasets consists of multiple articles, books, etc.)

- We also need to adjust the tokenizer accordingly so that it knows when and how to use the new `<unk>` token

````python
class SimpleTokenizerV2:
    """Same as SimpleTokenizerV1 but handles unknown tokens via the
    special "<|unk|>" marker, and supports the special "<|endoftext|>"
    token at decode time.

    Three methods to implement:

    __init__(self, vocab):
        Same as V1 -- store both the forward (string -> id) and inverse
        (id -> string) mappings on self.

    encode(self, text):
        Split and strip the text exactly as V1 did. Then, before doing
        the dictionary lookup, replace every token that is NOT already
        in the forward mapping with the literal string "<|unk|>". Look
        up the resulting list in the forward mapping and return the ids.

    decode(self, ids):
        Identical to V1 -- map ids back to strings, join with spaces,
        and clean up the punctuation spacing.
    """
    # YOUR CODE HERE
    def __init__(self, vocab):
        self.vocab = vocab
        self.string_to_integer = vocab
        self.integer_to_string = {i:s for s,i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()] 
        preprocessed = [
            item if item in self.string_to_integer else "<|unk|>"
            for item in preprocessed
        ]
        ids = [self.string_to_integer[s] for s in preprocessed]
        return ids
    
    def decode(self, ids):
        text = " ".join([self.integer_to_string[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'\\])', r'\1', text)
        return text

````

- We use a sliding window approach, changing the position by +1:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/13.webp?123" style="max-width: 100%; height: auto;">

- Create dataset and dataloader that extract chunks from the input text dataset

````python
from torch.utils.data import Dataset, DataLoader


class GPTDatasetV1(Dataset):
    """A sliding-window dataset for next-token prediction.

    Implement three methods on the class:

    __init__(self, txt, tokenizer, max_length, stride):
        Initialise two empty lists on self that will store the input
        chunks and the target chunks. Then tokenise the entire text in
        one call to the tokenizer's encode method, allowing the special
        "<|endoftext|>" token. Walk a sliding window of length
        max_length across the resulting list of token ids, advancing by
        `stride` positions each step. At each window position, take the
        current slice of length max_length as the input chunk and the
        slice shifted by one position to the right (also of length
        max_length) as the target chunk -- so each target token is the
        next-token prediction for its corresponding input token. Convert
        each pair to a torch tensor of longs and append them to the two
        lists.

    __len__(self):
        Return the number of (input, target) pairs you stored.

    __getitem__(self, idx):
        Return the (input_chunk, target_chunk) pair at index idx as a
        tuple of two tensors.
    """
    # YOUR CODE HERE
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []

        # initialization the tokenids
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})

        for i in range(0,len(token_ids) - max_length, stride):
            input_chunk = token_ids[i : i+max_length]
            target_chunk = token_ids[i+1 : i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids)
    
    def __getitem__(self, index):
        return self.input_ids[index], self.target_ids[index]

        
````

````python
def create_dataloader_v1(txt, batch_size=4, max_length=256,
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):
    """Build a DataLoader that yields (input_ids, target_ids) batches.

    Steps:
      - Instantiate the GPT-2 BPE tokenizer using tiktoken's get_encoding
        helper (the encoding name is "gpt2").
      - Wrap the raw text and the tokenizer in a GPTDatasetV1 instance,
        passing through max_length and stride from the function arguments.
      - Wrap that dataset in a PyTorch DataLoader using the batch_size,
        shuffle, drop_last, and num_workers arguments as configured.
      - Return the DataLoader.
    """
    # YOUR CODE HERE
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=shuffle, drop_last=drop_last, num_workers=num_workers)

    return dataloader
````

#### Adding token + positional embeddings (broadcasting)

To create the input embeddings, we simply add the token embeddings and the positional embeddings:

```
token_embeddings : (8, 4, 256)   — 8 sequences × 4 tokens × 256-dim vector
pos_embeddings   :    (4, 256)   — 4 positions × 256-dim vector
```

PyTorch resolves the shape mismatch using **broadcasting**: it aligns shapes right-to-left, treats the missing leading dimension of `pos_embeddings` as size 1, and stretches it to match the batch dimension of 8:

```
token_embeddings : (8, 4, 256)
pos_embeddings   : (1, 4, 256)  ← missing dim treated as 1, then broadcast to 8
result           : (8, 4, 256)
```

This means the **same** `pos_embeddings` is added to every sequence in the batch — which is exactly right, because "position 0" means *the first token* regardless of which sequence you are looking at. The positional embedding encodes *where*, not *what*, so it is batch-agnostic.

````python
input_embeddings = token_embeddings + pos_embeddings
print(input_embeddings.shape)

# uncomment & execute the following line to see how the embeddings look like
print(input_embeddings)
````
