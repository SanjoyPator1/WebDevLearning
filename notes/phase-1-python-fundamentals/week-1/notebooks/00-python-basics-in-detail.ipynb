## Lists

A list is a mutable, ordered collection of items in Python. Lists are one of the most versatile data structures and can store elements of different types (e.g., integers, strings, floats, or even other lists).

### Key Properties
1. **Ordered**: Elements in a list are stored in a specific order, and this order is preserved.
2. **Mutable**: Lists can be modified after their creation (e.g., adding, removing, or changing elements).
3. **Dynamic**: Lists can grow or shrink in size as needed.
Indexable: Elements can be accessed using zero-based indexing.

---



## List Operations: Creating, Slicing, Appending, and Extending

### Creating, Slicing, Appending, and Extending Lists


````python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Slicing
print(fruits[0:2])  # ['apple', 'banana']

# Appending
fruits.append("date")
print(fruits)  # ['apple', 'banana', 'cherry', 'date']

# Extending
fruits.extend(["elderberry", "fig"])
print(fruits)  # ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig']
````

### 1. Creating a List
A list is created using square brackets ([]), with items separated by commas.

````python
# Creating a list of fruits
fruits = ["apple", "banana", "cherry"]
print(fruits)  # Output: ['apple', 'banana', 'cherry']

#You can also create an empty list:
empty_list = []
print(empty_list)  # Output: []

# Lists can hold elements of different data types:
mixed_list = [1, "two", 3.0, True]
print(mixed_list)  # Output: [1, 'two', 3.0, True]

````

---
### 2. Slicing
Slicing is used to extract a portion of a list. The syntax is list[start:end], where:
- start: The index to begin the slice (inclusive).
- end: The index to stop the slice (exclusive).

If omitted:
- start defaults to 0.
- end defaults to the length of the list.

````python
# Slicing to extract the first two elements
print(fruits[0:2])  # Output: ['apple', 'banana']

# Omitting start and end
print(fruits[:2])  # Output: ['apple', 'banana']
print(fruits[1:])  # Output: ['banana', 'cherry']

# Using negative indexing
print(fruits[-2:])  # Output: ['banana', 'cherry']
````

---
### 3. Appending
The append() method adds a single element to the end of the list.

````python
# Appending an element to the list
fruits.append("date")
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'date']
````

- The list grows in size when an element is appended.
- Appending does not create a new list; it modifies the original one.


---
### 4. Extending
The extend() method is used to add multiple elements (from another iterable, such as a list) to the end of the list.

````python
# Extending the list with another list
fruits.extend(["elderberry", "fig"])
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig']
````

- Unlike append(), which adds the entire iterable as a single element, extend() unpacks the iterable and adds each item separately.

---
### Comparison Between append() and extend()
- append(): Adds the argument as a single element to the end of the list.

````python
fruits.append(["grape", "honeydew"])
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig', ['grape', 'honeydew']]
````

**Notice the nested list at the end.**



- extend(): Unpacks the argument (if it's an iterable) and adds each item individually.

````python
fruits.extend(["kiwi", "lemon"])
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig', ['grape', 'honeydew'], 'kiwi', 'lemon']
````

## Dictionaries
A dictionary is an unordered collection of key-value pairs. Each key in a dictionary is unique, and it maps to a corresponding value. Dictionaries are often used to store data that is logically associated, such as a name with a phone number or a username with an email address.

### Key Properties
1. **Key-Value Pairs**: Data is stored as key-value pairs, e.g., key: value.
2. **Keys Must Be Unique**: No two keys can have the same value in the same dictionary.
3. **Mutable**: Dictionaries can be updated (e.g., adding, removing, or changing key-value pairs).
4. **Unordered**: As of Python 3.7+, dictionaries maintain insertion order, but earlier versions do not guarantee this.


### Adding, Updating, and Deleting Key-Value Pairs

````python
# Create a dictionary
person = {"name": "Sanjoy", "age": 25}

# Adding a key-value pair
person["city"] = "Kolkata"

# Updating a value
person["age"] = 26

# Deleting a key-value pair
del person["city"]

print(person)  # {'name': 'Sanjoy', 'age': 26}
````

### Iterating through Keys and Values

````python
for key, value in person.items():
    print(key, ":", value)
````

---
### Detailed notes
## Dictionary Operations: Creating, Accessing, Adding, Updating, and Removing


1. **Creating a Dictionary**

  A dictionary is created using curly braces {} with keys and their corresponding values separated by a colon :.


````python
# Creating a dictionary
person = {"name": "Alice", "age": 30, "city": "New York"}
print(person)  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
````

You can also use the dict() function:

````python
person = dict(name="Alice", age=30, city="New York")
print(person)  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
````

2. **Accessing Values**

Values in a dictionary are accessed using their keys.

````python
# Accessing values by key
print(person["name"])  # Output: Alice
print(person["age"])   # Output: 30
````

- **KeyError**: Attempting to access a non-existent key will raise a KeyError. To avoid this, use the get() method, which returns None (or a default value) if the key doesn’t exist.


````python
print(person.get("job", "Not specified"))  # Output: Not specified
````

3. **Adding or Updating Key-Value Pairs**

You can add new key-value pairs or update existing ones by assigning a value to a key.

````python
# Adding a new key-value pair
person["job"] = "Engineer"
print(person)  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York', 'job': 'Engineer'}

# Updating an existing key's value
person["age"] = 31
print(person)  # Output: {'name': 'Alice', 'age': 31, 'city': 'New York', 'job': 'Engineer'}
````

---

4. **Removing Key-Value Pairs**
Dictionaries support multiple ways to remove elements:
- **del**: Deletes a key-value pair.
- **pop()**: Removes a key-value pair and returns its value.
- **popitem()**: Removes and returns the last inserted key-value pair (from Python 3.7+).


````python
# Using del
del person["city"]
print(person)  # Output: {'name': 'Alice', 'age': 31, 'job': 'Engineer'}

# Using pop
job = person.pop("job")
print(job)      # Output: Engineer
print(person)   # Output: {'name': 'Alice', 'age': 31}

# Using popitem
last_item = person.popitem()
print(last_item)  # Output: ('age', 31)
print(person)     # Output: {'name': 'Alice'}
````

5. **Iterating Through Dictionaries**

You can iterate over a dictionary to access its keys, values, or key-value pairs:

````python
# Iterating over keys
for key in person:
    print(key)  # Output: name

# Iterating over values
for value in person.values():
    print(value)  # Output: Alice

# Iterating over key-value pairs
for key, value in person.items():
    print(f"{key}: {value}")  # Output: name: Alice
````

6. **Dictionary Comprehension**

Dictionary comprehension allows you to construct dictionaries in a concise way:

````python
# Create a dictionary of squares
squares = {x: x**2 for x in range(5)}
print(squares)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
````

### Checking for Key Existence in a Dictionary
In Python, there are multiple ways to check whether a key exists in a dictionary. The most common and efficient methods are using the in keyword or the .get() method.



### 1. Using in Keyword
The in keyword checks if a key exists in the dictionary. This approach is both simple and efficient.

Syntax:
```python
key in dictionary
```

````python
# Define a dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Check if key 'b' exists
if 'b' in my_dict:
    print("Key 'b' is present.")
else:
    print("Key 'b' is not present.")

# Check for a key that is not present
if 'd' in my_dict:
    print("Key 'd' is present.")
else:
    print("Key 'd' is not present.")
````

### 2. Using .get() Method
The .get() method retrieves the value of a key if it exists. If the key is not found, it returns None by default or a specified default value.

Syntax:
```python
dictionary.get(key, default_value)
```


````python
# Using .get() to check for a key
value = my_dict.get('b')  # Returns 2
if value is not None:
    print("Key 'b' is present with value:", value)
else:
    print("Key 'b' is not present.")

# Specifying a default value
value = my_dict.get('d', "Key not found")
print(value)
````

### 3. Key-Value Pair Check with dict.keys()
You can use the .keys() method to explicitly check for keys, but this is usually unnecessary since in works directly with the dictionary.

Example:

````python
if 'b' in my_dict.keys():
    print("Key 'b' exists.")
````

This is equivalent to using in on the dictionary itself but less efficient.

### Practice Task
Write a program to check if a key exists in a dictionary. If it exists, print its value; otherwise, print "Key not found."

````python
my_dict = {'x': 10, 'y': 20, 'z': 30}

# Check for a key
key_to_check = 'y'
if key_to_check in my_dict:
    print(f"Key '{key_to_check}' exists with value: {my_dict[key_to_check]}")
else:
    print(f"Key '{key_to_check}' not found.")
````

### Comparison: `in` vs. `.get()`

| **Method**      | **Use Case**                                                                                         |
|------------------|-----------------------------------------------------------------------------------------------------|
| `in`            | Best for checking the presence of a key without needing its value.                                   |
| `.get()`         | Useful if you need the value of the key or want to specify a default if the key is absent.           |

## Tuples
A tuple is an immutable, ordered collection of elements. Unlike lists, tuples cannot be modified after they are created. They are typically used for storing related pieces of data that should not change during the lifetime of the program.


### Key Properties of Tuples
1. **Ordered**: Tuples maintain the order of elements, meaning that the items are stored in a specific sequence and can be accessed by their index.
2. **Immutable**: Once a tuple is created, you cannot add, remove, or modify its elements. This makes tuples useful for storing constant data.
3. **Allow Duplicates**: Just like lists, tuples can contain duplicate elements.
4. **Heterogeneous**: Tuples can store elements of different data types (e.g., integers, strings, lists).
6. **Can Contain Other Data Structures**: Tuples can hold other tuples, lists, or even dictionaries as elements.

### Creating and Accessing Tuples


### 1. Creating a Tuple
Tuples are created by placing comma-separated values inside parentheses (). You can also create a tuple without parentheses, simply by separating values with commas.

````python
# Creating a tuple with parentheses
my_tuple = (1, 2, 3, "apple", "banana")
print(my_tuple)  # Output: (1, 2, 3, 'apple', 'banana')

# Creating a tuple without parentheses (comma-separated values)
my_tuple2 = 1, 2, 3
print(my_tuple2)  # Output: (1, 2, 3)
````

### 2. Accessing Elements
You can access the elements of a tuple using indexing, just like in lists. Indexing starts from 0.


````python
# Accessing elements by index
print(my_tuple[0])  # Output: 1
print(my_tuple[3])  # Output: 'apple'

# Negative indexing (starts from the end)
print(my_tuple[-1])  # Output: 'banana'
````

### 3. Slicing a Tuple
Tuples support slicing, which allows you to get a subset of the tuple.


````python
# Slicing a tuple
print(my_tuple[1:4])  # Output: (2, 3, 'apple')

# Slicing with a step
print(my_tuple[::2])  # Output: (1, 3, 'banana')
````

## Tuple Operations


### 1. Concatenation
You can concatenate tuples using the + operator.

````python
# Concatenating tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined_tuple = tuple1 + tuple2
print(combined_tuple)  # Output: (1, 2, 3, 4, 5, 6)
````

### 2. Repeating Tuples
You can repeat the elements of a tuple using the * operator.

````python
# Repeating tuples
tuple3 = ("apple", "banana")
repeated_tuple = tuple3 * 2
print(repeated_tuple)  # Output: ('apple', 'banana', 'apple', 'banana')
````

### 3. Length of a Tuple
You can use the len() function to get the number of elements in a tuple.

````python
# Getting the length of a tuple
print(len(my_tuple))  # Output: 5
````

### 4. Checking for Membership
You can check if an element is present in a tuple using the in keyword.

````python
# Checking if an element is in the tuple
print("apple" in my_tuple)  # Output: True
print("orange" in my_tuple)  # Output: False
````

Once created, the elements in a tuple cannot be modified (i.e., you cannot change, add, or remove elements). This is in contrast to lists, which are mutable.

## Nested Tuples
Tuples can contain other tuples, lists, or even dictionaries as elements.

````python
# Tuple containing another tuple
nested_tuple = (1, (2, 3), ("apple", "banana"))
print(nested_tuple)  # Output: (1, (2, 3), ('apple', 'banana'))
print(nested_tuple[1])  # Output: (2, 3)
print(nested_tuple[2][0])  # Output: 'apple'
````

## Use Cases of Tuples
- **Data Integrity**: When you need a collection of values that shouldn't be modified, tuples offer a way to maintain data integrity.
- **Heterogeneous Data**: Tuples are often used for storing heterogeneous data (e.g., a person’s name, age, and address).
- **Efficient Memory Use**: Since tuples are immutable, they are more memory efficient compared to lists and can be faster for certain use cases.
- **Return Multiple Values**: Functions that return multiple values can use tuples to return a set of values.

## Code Example: Using Tuples

````python
# Creating a tuple
student = ("Alice", 21, "Math", 3.9)

# Accessing tuple elements
name = student[0]
age = student[1]
major = student[2]
gpa = student[3]

# Printing values
print(f"Name: {name}, Age: {age}, Major: {major}, GPA: {gpa}")

# Nested tuple example
coordinates = (40.7128, 74.0060)  # (latitude, longitude)
print(f"Latitude: {coordinates[0]}, Longitude: {coordinates[1]}")
````

## Tuple Unpacking
You can unpack the elements of a tuple into variables.

````python
# Tuple unpacking
person = ("John", 30, "Engineer")
name, age, occupation = person
print(f"Name: {name}, Age: {age}, Occupation: {occupation}")
````

## Advantages of Tuples

2. **Memory Efficiency**: Tuples consume less memory than lists because they are immutable.
3. **Faster than Lists**: Since they are immutable, they are generally faster than lists for iteration and accessing elements.

- Use tuples to return multiple values from a function.

## Sets
A set is an unordered collection of unique elements. Sets are primarily used to perform operations like union, intersection, and difference efficiently. They are particularly useful when you need to eliminate duplicate items or check membership.

## Key Properties of Sets
1. **Unordered**: Elements in a set are not stored in a specific order and cannot be accessed using an index.
2. **Unique Elements**: Sets automatically remove duplicate values.
3. **Mutable**: You can add or remove elements from a set after its creation.
4. **Heterogeneous**: A set can store elements of different data types.
5. **Non-hashable Items Not Allowed**: Sets cannot contain mutable or non-hashable items like lists, dictionaries, or other sets.

## Creating and Accessing Sets


### 1. Creating a Set
Sets are created using curly braces {} or the set() constructor.

````python
# Creating a set with curly braces
my_set = {1, 2, 3, 4}
print(my_set)  # Output: {1, 2, 3, 4}

# Creating a set using the set() constructor
my_set2 = set([1, 2, 2, 3])
print(my_set2)  # Output: {1, 2, 3}

# Creating an empty set (use set(), not {})
empty_set = set()
print(type(empty_set))  # Output: <class 'set'>
````

### 2. Accessing Elements
Since sets are unordered, elements cannot be accessed using indexing or slicing. However, you can loop through a set to access elements.

````python
# Looping through a set
for element in my_set:
    print(element)
````

---
## Set Operations

### 1. Adding Elements
You can add elements to a set using the add() method.

````python
# Adding an element to a set
my_set.add(5)
print(my_set)  # Output: {1, 2, 3, 4, 5}
````

### 2. Removing Elements
You can remove elements using the remove() or discard() methods. If the element does not exist, remove() raises an error, while discard() does not.

````python
# Removing an element
my_set.remove(3)
print(my_set)  # Output: {1, 2, 4}

# Using discard (no error if element not present)
my_set.discard(6)  # No error
````

### 3. Clearing a Set
You can remove all elements using the clear() method.

````python
# Clearing a set
my_set.clear()
print(my_set)  # Output: set()
````

---
## Mathematical Set Operations


### 1. Union
The union() method or | operator combines all unique elements from two sets.

````python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.union(set2))  # Output: {1, 2, 3, 4, 5}
print(set1 | set2)       # Output: {1, 2, 3, 4, 5}
````

### 2. Intersection
The intersection() method or & operator returns elements common to both sets.

````python
print(set1.intersection(set2))  # Output: {3}
print(set1 & set2)             # Output: {3}
````

### 3. Difference
The difference() method or - operator returns elements in the first set but not in the second.

````python
print(set1.difference(set2))  # Output: {1, 2}
print(set1 - set2)           # Output: {1, 2}
````

### 4. Symmetric Difference
The symmetric_difference() method or ^ operator returns elements in either set but not in both.

````python
print(set1.symmetric_difference(set2))  # Output: {1, 2, 4, 5}
print(set1 ^ set2)                      # Output: {1, 2, 4, 5}
````

---
## Other Set Methods

### 1. Checking Membership
You can check if an element is in a set using the in keyword.

````python
print(2 in set1)  # Output: True
print(6 in set1)  # Output: False
````

### 2. Subset and Superset
You can check if one set is a subset or superset of another using issubset() and issuperset().

````python
set3 = {1, 2}
print(set3.issubset(set1))  # Output: True
print(set1.issuperset(set3))  # Output: True
````

### 3. Copying a Set
You can copy a set using the copy() method.

````python
set_copy = set1.copy()
print(set_copy)  # Output: {1, 2, 3}
````

---
## Applications of Sets
1. **Removing Duplicates**: Sets are commonly used to eliminate duplicate elements from a list or other iterable.
2. **Membership Testing**: Sets provide efficient ways to check if an element exists in a collection.
3. **Mathematical Operations**: Union, intersection, and difference are useful in various scenarios, such as finding common or unique items between two datasets.

## Code Example: Practical Use of Sets

````python
# Removing duplicates from a list
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = list(set(numbers))
print(unique_numbers)  # Output: [1, 2, 3, 4, 5]

# Finding common elements between two lists
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
common_elements = set(list1) & set(list2)
print(common_elements)  # Output: {3, 4}

# Checking if all elements of one list are in another
set1 = {1, 2, 3}
set2 = {1, 2}
print(set2.issubset(set1))  # Output: True
````

## Advantages of Sets
1. **Unique Elements**: Automatically removes duplicates.
2. **Efficient Membership Testing**: Faster lookups compared to lists.
3. **Mathematical Operations**: Built-in methods for union, intersection, and difference make set manipulation intuitive.


## When to Use Sets
- Use sets when you need to store unique elements.
- Use sets for efficient membership testing.
- Use sets for mathematical operations like union, intersection, and difference.

## Iterators and Generators
### Key Concepts
1. Iterators:
  - Objects that implement the __iter__() and __next__() methods.
  - Used to iterate over sequences like lists, tuples, etc.
2. Generators:
  - Created using functions with the yield statement.
  - Produce items one at a time and are memory efficient.
  

---
### Examples
### Custom Iterator

````python
# Custom Iterator to iterate over a range of numbers
class MyIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.end:
            raise StopIteration
        val = self.current
        self.current += 1
        return val

# Using the iterator
it = MyIterator(1, 5)
for num in it:
    print(num)  # Output: 1 2 3 4
````

### Generator for Fibonacci Numbers

````python
# Fibonacci Generator
def fibonacci(n):
    a, b = 0, 1
    while a <= n:
        yield a
        a, b = b, a + b

# Using the generator
for num in fibonacci(10):
    print(num)  # Output: 0 1 1 2 3 5 8
````

### Decorators
### Key Concepts
- A decorator is a function that takes another function as input and returns a modified version.
- Commonly used for logging, access control, or measuring execution time.
---
#### Examples
##### Simple Decorator

````python
# Example: Logging decorator
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with arguments {args} and {kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(3, 5))  # Output: Logs and prints the result 8
````

#### Decorator for Measuring Execution Time

````python
import time

# Example: Timer decorator
def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time:.2f} seconds")
        return result
    return wrapper

@timer
def compute_sum(n):
    return sum(range(n))

print(compute_sum(1000000))  # Output: Execution time and result
````

## Regular Expressions
Regular expressions (regex) are used for pattern matching within strings. Python’s re module provides functions for matching, searching, and replacing patterns in strings.

### Key Concepts
1. Basic Syntax:
  - . : Matches any character except newline.
  - $[ ]$ : Matches any one character inside the brackets.
  - ^ : Matches the beginning of a string.
  - $ : Matches the end of a string.

  - $ * $ : Matches 0 or more occurrences of the preceding - element.
  - $+$ : Matches 1 or more occurrences of the preceding - element.
  - {n} : Matches exactly n occurrences of the preceding - element.
  - | : Matches either of two options.
  - () : Groups expressions together.
2. Functions in re module:
  - re.match(): Checks if the pattern matches at the start of the string.
  - re.search(): Searches for the pattern anywhere in the string.
  - re.findall(): Finds all occurrences of the pattern.
  - re.sub(): Replaces the occurrences of the pattern with a specified string.

---
### Examples
#### Matching Email Addresses

````python
import re

# Sample string
text = "Please contact us at support@example.com or sales@company.org"

# Regex to match email addresses
emails = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', text)
print(emails)  # Output: ['support@example.com', 'sales@company.org']
````

#### Using re.sub() to Replace Text

````python
# Replacing 'dog' with 'cat'
text = "The dog is in the house."
result = re.sub(r'dog', 'cat', text)
print(result)  # Output: The cat is in the house.
````
